```python
import os
import requests
import zipfile
```

```python
# define file paths
home = r'C:\Users\carroll\Documents\sbgPlant'

col = os.path.join(home, 'data', 'col_2018', 'raw')
os.makedirs(col, exist_ok=True)

dois = {
    os.path.join(col, '10.15485.1618130.zip'): 'https://data.ess-dive.lbl.gov/catalog/d1/mn/v2/packages/application%2Fbagit-1.0/ess-dive-97f8b9398482301-20230504T205732518573',
    os.path.join(col, '10.15485.1618132.zip'): 'https://data.ess-dive.lbl.gov/catalog/d1/mn/v2/packages/application%2Fbagit-1.0/ess-dive-e1f7ade605f0d02-20230504T210959436515',
    os.path.join(col, '10.15485.1631278.zip'): 'https://data.ess-dive.lbl.gov/catalog/d1/mn/v2/packages/application%2Fbagit-1.0/ess-dive-255747bbf26d08c-20230504T211005800913'
}
```

```python
# function to download per doi
def download(fp_out, url):
    response = requests.get(url, stream=True)
    if response.status_code == 200:
        with open(fp_out, 'wb') as f:
            for chunk in response.iter_content(chunk_size=8192):
                f.write(chunk)
    else:
        print(f'failed - doi:{doi}: Status {response.status_code}')
```

```python
for fp_out, url in dois.items():
    download(fp_out, url)
```

```python
for fp_zip, url in dois.items():
    out_dir = fp_zip.strip('.zip')
    os.makedirs(out_dir, exist_ok=True)
    with zipfile.ZipFile(fp_zip, 'r') as zip_ref:
        zip_ref.extractall(out_dir)
    os.remove(fp_zip)
```
