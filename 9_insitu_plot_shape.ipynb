```python
import os
import pandas as pd
from glob import glob
import numpy as np
import geopandas as gpd

os.chdir('/store/carroll/sbgplants/')
```

```python
# file paths
raw = 'data/raw'

doi = os.path.join(raw, '10.15485.1618130') # Locations, metadata, and species cover from field sampling survey associated with NEON AOP survey, East River, CO 2018

out_folder = 'data/out_csv'

table = 'insitu_plot_shape'
```

```python
# load relevant data

shapes = gpd.read_file(os.path.join(doi, 'CRBU2018_AOP_Crowns.geojson'))
insitu_plot = pd.read_csv(os.path.join(out_folder, 'insitu_plot_event.csv'))[['plot_name','insitu_plot_event_id','collection_date']]
```

```python
# prepare & populate out table

out_table = shapes
out_table = out_table.drop(columns='id')

out_table = out_table.merge(insitu_plot, left_on='SiteCode', right_on='plot_name', how='left', suffixes=('',''))
out_table['insitu_plot_shape_id'] = range(len(out_table))

out_table = out_table[['insitu_plot_shape_id', 'collection_date', 'insitu_plot_event_id', 'geometry']]

out_table
```

```python
# export table
fp_out = os.path.join(out_folder, f'{table}.geojson')
out_table.to_file(fp_out, index=False)
```
