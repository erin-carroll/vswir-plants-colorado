```python
import os
import pandas as pd
import random
import numpy as np
```

```python
rdn = pd.read_csv(r"C:\Users\carroll\Documents\sbgplant\extracted_spectra.csv")
loc = pd.read_csv(r"C:\Users\carroll\Documents\sbgplant\extracted_locations.csv")
obs = pd.read_csv(r"C:\Users\carroll\Documents\sbgplant\extracted_observations.csv")

rdn = rdn.pivot(index='pixel_id', columns='band_number', values='radiance').reset_index(drop=False)
loc = loc.pivot(index='pixel_id', columns='loc_type', values='loc_value').reset_index(drop=False)
obs = obs.pivot(index='pixel_id', columns='obs_type', values='obs_value').reset_index(drop=False)
```

```python
pixel = pd.read_csv(r"C:\Users\carroll\Documents\sbgplant\pixel.csv")[['pixel_id', 'granule_id']]
```

```python
df = rdn.merge(loc, on='pixel_id', how='left').merge(obs, on='pixel_id', how='left').merge(pixel, on='pixel_id', how='left')
```

```python
df['sensor_name'] = 'NIS01'
df['campaign_name'] = 'CRBU 2018'
df['acquisition_start_time'] = [x.split('_')[-1] for x in df['granule_id']]
df['acquisition_date'] = [x.split('_')[1] for x in df['granule_id']]
```

```python
fp = r"C:\Users\carroll\Documents\sbgplant\plants_isofit_extraction_crbu2018.csv"
df.to_csv(fp, index=False)
```
